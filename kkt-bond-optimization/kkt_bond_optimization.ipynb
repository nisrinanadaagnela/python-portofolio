# Modal Investasi: 30.000 DM

Code cell di bawah ini memberikan solusi untuk soal nomor 1 dan 2

```python
import cvxpy as cp

# Masukkan data
V = {5: 150.01, 6: 163.51, 7: 178.23}
initial_investment = 30000
tax_free_limit = 6100
tax_rate = 0.3

# Mendefinisikan variabel
x = {t: cp.Variable(nonneg=True) for t in [5, 6, 7]}

# Definisikan constraints
constraints = [
    x[5] + x[6] + x[7] == 1,  # Total proportion constraint
    x[5] >= 0, x[5] <= 1,
    x[6] >= 0, x[6] <= 1,
    x[7] >= 0, x[7] <= 1
]

# Fungsi objektif
total_gain = 0
for t in [5, 6, 7]:
    I_t = x[t] * initial_investment * (V[t] / 100 - 1)
    S_t = initial_investment * (1 - x[t]) * (V[t] / 100 - 1)
    P_t = tax_rate * cp.pos(S_t - tax_free_limit)
    S_t = S_t - P_t
    total_gain += I_t + S_t

# Maximize the total return
objective = cp.Maximize(total_gain)

# Problem definition
problem = cp.Problem(objective, constraints)

# Check if the objective function is convex
is_convex = objective.is_dcp()
print(f"Is the objective function convex? {'Yes' if is_convex else 'No'}")

# Solve the problem
problem.solve()

# Results
print(f"\nOptimal value: {problem.value}")
print(f"Proportion of bonds sold at the end of year 5: {round(x[5].value, 3)}")
print(f"Proportion of bonds sold at the end of year 6: {round(x[6].value, 3)}")
print(f"Proportion of bonds sold at the end of year 7: {round(x[7].value, 3)}")

# Verify KKT conditions
# Stationarity: grad_f + sum(lambda_i * grad_g_i) = 0
lambdas = [con.dual_value for con in constraints]

grad_f = []
for t in [5, 6, 7]:
    I_t_grad = initial_investment * (V[t] / 100 - 1)
    S_t_grad = -initial_investment * (V[t] / 100 - 1)
    grad_f.append(I_t_grad + S_t_grad)

stationarity = []
for t in range(3):
    stationarity.append(grad_f[t] + lambdas[0])  # Lambda untuk kendala total proporsi

# Print Lagrange multipliers
print("\nLagrange Multipliers:")
for i, con in enumerate(constraints):
    print(f"Lambda for constraint {i+1}: {con.dual_value}")
```

```python
# Sertifikat Deposito
import cvxpy as cp

# Masukkan data
V = {5: 150.01, 6: 163.51, 7: 178.23}
initial_investment = 30000
tax_free_limit = 6100
tax_rate = 0.3
deposit_interest_rate = 0.04  # 4% annual interest rate for the certificates of deposit

# Mendefinisikan variabel
x = {t: cp.Variable(nonneg=True) for t in [5, 6, 7]}

# Definisikan constraints
constraints = [
    x[5] + x[6] + x[7] == 1,  # Total proportion constraint
    x[5] >= 0, x[5] <= 1,
    x[6] >= 0, x[6] <= 1,
    x[7] >= 0, x[7] <= 1
]

# Fungsi objektif
total_gain = 0
for t in [5, 6, 7]:
    I_t = x[t] * initial_investment * (V[t] / 100 - 1)
    S_t = initial_investment * (1 - x[t]) * (V[t] / 100 - 1)
    P_t = tax_rate * cp.pos(S_t - tax_free_limit)
    S_t = S_t - P_t
    deposit_interest = (1 - x[t]) * initial_investment * deposit_interest_rate
    total_gain += I_t + S_t + deposit_interest

# Maximize the total return
objective = cp.Maximize(total_gain)

# Problem definition
problem = cp.Problem(objective, constraints)

# Check if the objective function is convex
is_convex = objective.is_dcp()
print(f"Is the objective function convex? {'Yes' if is_convex else 'No'}")

# Solve the problem
problem.solve()

# Results
print(f"\nOptimal value: {problem.value}")
print(f"Proportion of bonds sold at the end of year 5: {round(x[5].value, 3)}")
print(f"Proportion of bonds sold at the end of year 6: {round(x[6].value, 3)}")
print(f"Proportion of bonds sold at the end of year 7: {round(x[7].value, 3)}")

# Verify KKT conditions
# Stationarity: grad_f + sum(lambda_i * grad_g_i) = 0
lambdas = [con.dual_value for con in constraints]

grad_f = []
for t in [5, 6, 7]:
    I_t_grad = initial_investment * (V[t] / 100 - 1)
    S_t_grad = -initial_investment * (V[t] / 100 - 1)
    grad_f.append(I_t_grad + S_t_grad)

stationarity = []
for t in range(3):
    stationarity.append(grad_f[t] + lambdas[0])  # Lambda untuk kendala total proporsi

# Print Lagrange multipliers
print("\nLagrange Multipliers:")
for i, con in enumerate(constraints):
    print(f"Lambda for constraint {i+1}: {con.dual_value}")
```

```python
# Kalau Charles menikah
import cvxpy as cp

# Masukkan data
V = {5: 150.01, 6: 163.51, 7: 178.23}
initial_investment = 30000
tax_free_limit_5 = 6100
tax_free_limit_67 = 12200
tax_rate = 0.3

# Mendefinisikan variabel
x = {t: cp.Variable(nonneg=True) for t in [5, 6, 7]}

# Definisikan constraints
constraints = [
    x[5] + x[6] + x[7] == 1,  # Total proportion constraint
    x[5] >= 0, x[5] <= 1,
    x[6] >= 0, x[6] <= 1,
    x[7] >= 0, x[7] <= 1
]

# Fungsi objektif
total_gain = 0
for t in [5, 6, 7]:
    I_t = x[t] * initial_investment * (V[t] / 100 - 1)
    S_t = initial_investment * (1 - x[t]) * (V[t] / 100 - 1)
    if t == 5:
        P_t = tax_rate * cp.pos(S_t - tax_free_limit_5)
    else:
        P_t = tax_rate * cp.pos(S_t - tax_free_limit_67)
    S_t = S_t - P_t
    total_gain += I_t + S_t

# Maximize the total return
objective = cp.Maximize(total_gain)

# Problem definition
problem = cp.Problem(objective, constraints)

# Solve the problem
problem.solve()

# Results
optimal_value = problem.value
x5_value = round(x[5].value, 3)
x6_value = round(x[6].value, 3)
x7_value = round(x[7].value, 3)

# Check if the objective function is convex
is_convex = objective.is_dcp()
print(f"Is the objective function convex? {'Yes' if is_convex else 'No'}")

print(f"\nOptimal value: {optimal_value}")
print(f"Proportion of bonds sold at the end of year 5: {x5_value}")
print(f"Proportion of bonds sold at the end of year 6: {x6_value}")
print(f"Proportion of bonds sold at the end of year 7: {x7_value}")

# Perhitungan penghematan
def calculate_savings(old_limit, new_limit):
    old_total_gain = 0
    new_total_gain = 0
    for t in [5, 6, 7]:
        I_t = x[t].value * initial_investment * (V[t] / 100 - 1)
        S_t = initial_investment * (1 - x[t].value) * (V[t] / 100 - 1)
        P_t_old = tax_rate * max(S_t - old_limit, 0)
        P_t_new = tax_rate * max(S_t - new_limit, 0)
        old_total_gain += I_t + (S_t - P_t_old)
        new_total_gain += I_t + (S_t - P_t_new)
    return old_total_gain, new_total_gain

old_gain, new_gain = calculate_savings(tax_free_limit_5, tax_free_limit_67)
savings = new_gain - old_gain

print(f"\nSavings due to increased tax-free limit: {savings}")

# Verify KKT conditions
lambdas = [con.dual_value for con in constraints]

grad_f = []
for t in [5, 6, 7]:
    I_t_grad = initial_investment * (V[t] / 100 - 1)
    S_t_grad = -initial_investment * (V[t] / 100 - 1)
    grad_f.append(I_t_grad + S_t_grad)

stationarity = []
for t in range(3):
    stationarity.append(grad_f[t] + lambdas[0])  # Lambda untuk kendala total proporsi

# Print Lagrange multipliers
print("\nLagrange Multipliers:")
for i, con in enumerate(constraints):
    print(f"Lambda for constraint {i+1}: {con.dual_value}")
```
