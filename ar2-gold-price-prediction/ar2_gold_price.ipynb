```python
import optuna
from pmdarima.arima import ARIMA
from sklearn.metrics import mean_absolute_percentage_error

def objective(trial):
    # Menentukan rentang pencarian parameter ARIMA
    p = trial.suggest_int('p', 0, 5)
    d = trial.suggest_int('d', 0, 2)
    q = trial.suggest_int('q', 0, 5)

    # Membuat model ARIMA dengan parameter yang dipilih
    model = ARIMA(order=(p, d, q))

    # Melatih model ARIMA
    model.fit(df['IDR'])

    # Membuat prediksi dengan model
    y_pred = model.predict_in_sample()
    y_true = df['IDR']

    # Menghitung MAPE dari model dan mengubah ke persen
    mape = mean_absolute_percentage_error(y_true, y_pred) * 100

    return mape

# Optimasi parameter menggunakan Optuna
study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=20)  # Menggunakan 100 iterasi pencarian

# Mendapatkan parameter terbaik yang ditemukan
best_params = study.best_params
print("Best Parameters:", best_params)

# Membuat model ARIMA dengan parameter terbaik
best_model = ARIMA(order=(best_params['p'], best_params['d'], best_params['q']))
best_model.fit(df['IDR'])

# Melihat model terbaik yang ditemukan
print("MAPE: {:.2f}%".format(study.best_value))
print(best_model.summary())
```

```python
from scipy import stats
import pandas as pd
import numpy as np

# DataFrame untuk menyimpan nilai skewness
skewness_scores = pd.DataFrame()

for col in df_train.select_dtypes(include=np.float64).columns:
    if col != "Euro":
        df_orig = df_train[col].dropna()

        skew_orig = stats.skew(df_orig)

        transformed_yeojohn, _ = stats.yeojohnson(df_orig)
        skew_yeojohn = stats.skew(transformed_yeojohn)

        df_log = np.log1p(df_orig)
        skew_log = stats.skew(df_log)

        df_sqrt = np.sqrt(df_orig)
        skew_sqrt = stats.skew(df_sqrt)

        try:
            transformed, _ = stats.boxcox(df_orig)
            skew_boxcox = stats.skew(transformed)
        except ValueError:
            skew_boxcox = np.nan

        skewness_scores[col] = [
            skew_orig,
            skew_yeojohn,
            skew_log,
            skew_sqrt,
            skew_boxcox
        ]
    else:
        skewness_scores[col] = [np.nan] * 5

# Menetapkan indeks untuk skewness_scores
skewness_scores.index = ["Original", "YeoJohnson", "Log", "Sqrt", "BoxCox"]
skewness_scores = skewness_scores.transpose()

# Menentukan transformasi terbaik berdasarkan skewness minimum absolut
skewness_scores["Skewness_Abs_Min"] = skewness_scores.abs().idxmin(axis=1)
```

```python
import optuna
import numpy as np
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_percentage_error

# Membagi data menjadi set pelatihan dan pengujian (80% train, 20% test)
train_size = int(len(df_train) * 0.8)
train, test = df_train.iloc[:train_size], df_train.iloc[train_size:]

def objective(trial):
    # Menentukan rentang pencarian parameter ARIMA
    p = trial.suggest_int('p', 0, 5)
    d = trial.suggest_int('d', 0, 2)
    q = trial.suggest_int('q', 0, 5)

    # Membuat model ARIMA dengan parameter yang dipilih
    model = ARIMA(train['diff_idr_1'], order=(p, d, q))

    # Melatih model ARIMA
    model_fit = model.fit()

    # Prediksi pada set pengujian
    forecast_diff = model_fit.forecast(steps=len(test))

    # Mengubah differencing menjadi nilai prediksi IDR asli
    inverted_diff = train['IDR'].iloc[-1] + np.cumsum(forecast_diff)

    # Menghitung MAPE dari model pada set pengujian dan mengubah ke persen
    mape = mean_absolute_percentage_error(test['IDR'].values, inverted_diff.values) * 100

    return mape

# Optimasi parameter menggunakan Optuna
study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=20)

# Mendapatkan parameter terbaik yang ditemukan
best_params = study.best_params
print("Best Parameters:", best_params)

# Membuat model ARIMA dengan parameter terbaik
best_model = ARIMA(train['diff_idr_1'], order=(best_params['p'], best_params['d'], best_params['q']))
best_model_fit = best_model.fit()

# Prediksi pada set pengujian dengan model terbaik
best_forecast_diff = best_model_fit.forecast(steps=len(test))

# Mengubah differencing menjadi nilai prediksi IDR asli
best_inverted_diff = train['IDR'].iloc[-1] + np.cumsum(best_forecast_diff)

# Menghitung MAPE pada set pengujian dengan model terbaik
best_mape_value = mean_absolute_percentage_error(test['IDR'].values, best_inverted_diff.values) * 100

print(f"MAPE: {best_mape_value:.2f}%")
print(best_model_fit.summary())
```

```python
import optuna
import numpy as np
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_percentage_error

# Membagi data menjadi set pelatihan dan pengujian (80% train, 20% test)
train_size = int(len(df_train) * 0.8)
train, test = df_train.iloc[:train_size], df_train.iloc[train_size:]

def objective(trial):
    # Menentukan rentang pencarian parameter ARIMA
    p = trial.suggest_int('p', 0, 5)
    d = trial.suggest_int('d', 0, 2)
    q = trial.suggest_int('q', 0, 5)

    # Membuat model ARIMA dengan parameter yang dipilih
    model = ARIMA(train['diff_idr_2'], order=(p, d, q))

    # Melatih model ARIMA
    model_fit = model.fit()

    # Prediksi pada set pengujian
    forecast_diff = model_fit.forecast(steps=len(test))

    # Mengubah differencing menjadi nilai prediksi IDR asli
    inverted_diff = train['IDR'].iloc[-1] + np.cumsum(forecast_diff)

    # Menghitung MAPE dari model pada set pengujian dan mengubah ke persen
    mape = mean_absolute_percentage_error(test['IDR'].values, inverted_diff.values) * 100

    return mape

# Optimasi parameter menggunakan Optuna
study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=20)

# Mendapatkan parameter terbaik yang ditemukan
best_params = study.best_params
print("Best Parameters:", best_params)

# Membuat model ARIMA dengan parameter terbaik
best_model = ARIMA(train['diff_idr_2'], order=(best_params['p'], best_params['d'], best_params['q']))
best_model_fit = best_model.fit()

# Prediksi pada set pengujian dengan model terbaik
best_forecast_diff = best_model_fit.forecast(steps=len(test))

# Mengubah differencing menjadi nilai prediksi IDR asli
best_inverted_diff = train['IDR'].iloc[-1] + np.cumsum(best_forecast_diff)

# Menghitung MAPE pada set pengujian dengan model terbaik
best_mape_value = mean_absolute_percentage_error(test['IDR'].values, best_inverted_diff.values) * 100

print(f"MAPE: {best_mape_value:.2f}%")
print(best_model_fit.summary())
```
