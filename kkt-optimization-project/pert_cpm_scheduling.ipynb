```python
def jalur_proyek_waktu_terlama(p):
  p1 = p
  for k in reversed(p1.keys()):
    pred = p1[k]["pred"]

    if p1[k]['pred'][0] == "END": # tidak ada predecessor
        p1[k]['start'] = p1[k]['end'] - p1[k]['duration']

    elif len(pred) == 1: # 1 predecessor
        key = p1[k]["pred"][0]

        p1[k]['end'] = p1[key]['start']
        p1[k]['start'] = p1[k]['end'] - p1[k]['duration']

    elif len(pred) > 1: # lebih dari 1 predecessor
        starts = [p1[p.strip()]['start'] for p in pred]
        p1[k]['end'] = min(starts)
        p1[k]['start'] = p1[k]['end'] - p1[k]['duration']

  return p1
```

```python
from pertchart import PertChart

pc = PertChart()
pc.create_pert_chart(jalur_proyek_waktu_terlama(proyek_uas_2))
```

```python
def jalur_kritis(data1, data2):
  slacks = []
  for k in data1.keys():
    slack = data2[k]['end'] - data1[k]['end']
    slacks.append(slack)
    if slack == 0:
      status = 'dalam jalur kritis'
    else:
      status = 'tidak dalam jalur kritis'
    print(f"Aktifitas {k} {status}")
  return slacks
```
